```python
class Usuario:

    banco = 'Bradesco' # atributo de classe, todo usuario vai pertencer ao banco 'Bradesco'

    def __init__(self, nome):
        self.nome = nome # atributo de instancia, cada usuario vai ter um nome diferente
    
    def info(self):
        print(self.nome)
        print(self.banco)

usuario1 = Usuario('Felipe')
usuario2 = Usuario('Renato')

usuario1.info()
usuario2.info()
```

```python
class TV:

    vol_min = 1
    vol_max = 5
    canal_min = 1
    canal_max = 6

    def __init__(self, canal=1, volume=2):
        self.ligado = False
        self.canal = canal
        self.volume = volume

    def ligar(self):
        self.ligado = True

    def passar_canal(self):
        canal += 1

    def voltar_canal(self):
        canal -= 1

    def aumentar_volume(self):
        self.volume += 1

    def diminuir_volume(self):
        self.volume -= 1

    def mostrar_tv(self):
        if not self.ligado:
            print('A tv está desligada! ')
        else:
            print('A tv está ligada.')
            print(f'Ch{self.canal} vol:{('0'*self.volume)+('/'*(self.vol_max-self.volume))}')

tv = TV(3, 5)
tv.mostrar_tv()
tv.ligar()
tv.mostrar_tv()
tv.voltar_canal()
```
