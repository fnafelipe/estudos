```python
class Cachorro:
    def __init__(self, nome, raca):
        self.nome = nome
        self.raca = raca
    
    def latir(self):
        print(f'O {self.raca} {self.nome} está latindo, AU AU AU!!!')

cachorro1 = Cachorro('Rex', 'Bulldog')
cachorro2 = Cachorro('Lino', 'Pinscher')

cachorro1.latir()
cachorro2.latir()
```

```python
'''
Celular
    atributos:
        cor
        memoria
        marca
        modelo
        programas
        tema
    metodos:
        tirar foto
        baixar app
        baixar jogo
        mostrar programas
        mudar tema
    a ideia é  a memoria diminuir a cada app (jogo ou app) ou foto
'''

class Celular:

    def __init__(self):
        self.cor = input('Cor: ')
        self.memoria = int(input('Memoria: '))
        self.marca = input('Marca: ')
        self.modelo = input('Modelo: ')
        self.programas = []
        self.tema = 'padrão'

    def mostrar_info(self):
        print(self.cor)
        print(self.memoria)
        print(self.marca)
        print(self.modelo)
        print(self.programas)
        print(self.tema)

    def tirar_foto(self):
        input('Pressione enter para tirar a foto ')
        self.memoria -= 1
        print('Foto de 1 MB salva.')
        print(f'Memoria atual: {self.memoria}')

    def baixar_app(self):
        app = input('App: ')
        self.programas.append(app)
        self.memoria -= 3
        print(f'App {app} baixado.')
        print(f'Memoria atual: {self.memoria}')

    def baixar_jogo(self):
        jogo = input('Jogo: ')
        self.programas.append(jogo)
        self.memoria -= 5
        print(f'Jogo {jogo} baixado.')
        print(f'Memoria atual: {self.memoria}')

    def mostrar_programas(self):
        print(self.programas)

    def mudar_tema(self):
        print(f'Atual: {self.tema}')
        tema = input('Novo: ')
        self.tema = tema
        print('Tema atualizado.')

celular = Celular()
celular.mostrar_info()
celular.tirar_foto()
celular.tirar_foto()
celular.baixar_app()
celular.baixar_jogo()
celular.mostrar_programas()
celular.mudar_tema()
celular.mostrar_info()
```
